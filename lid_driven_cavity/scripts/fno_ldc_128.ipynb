````python
import numpy as np

# 1. Define the base path to the dataset folder
path = "/content/sample_data/LDC_NS_2D/128x128/"

# 2. Create a list of the dataset variants
variants = ["harmonics", "nurbs", "skelneton"]

# 3. Initialize two empty lists for input and output data
all_X_data = []
all_Y_data = []

# 4. Iterate through each variant and load the data
for name in variants:
    # a. Construct the full file path for the input data
    x_file_path = f"{path}/{name}_lid_driven_cavity_X.npz"
    # b. Load the 'data' array from the input .npz file
    X_data = np.load(x_file_path)['data']
    all_X_data.append(X_data)

    # c. Construct the full file path for the output data
    y_file_path = f"{path}/{name}_lid_driven_cavity_Y.npz"
    # d. Load the 'data' array from the output .npz file
    Y_data = np.load(y_file_path)['data']
    all_Y_data.append(Y_data)

print("Loaded input (X) data for all variants. Number of datasets:", len(all_X_data))
print("Loaded output (Y) data for all variants. Number of datasets:", len(all_Y_data))
print(f"Example X data shape for '{variants[0]}': {all_X_data[0].shape}")
print(f"Example Y data shape for '{variants[0]}': {all_Y_data[0].shape}")
````

````python
X_combined = np.concatenate(all_X_data, axis=0)
Y_combined = np.concatenate(all_Y_data, axis=0)

print(f"Combined X data shape: {X_combined.shape}")
print(f"Combined Y data shape: {Y_combined.shape}")
````

````python
import torch

# --- 1. Normalize X_combined ---
print("Normalizing X_combined...")

X_normalized = np.copy(X_combined).astype(np.float32)

# Channel 0 (Reynolds number)
# It's a constant value per sample. Find unique values and their max for global scaling.
# The first dimension is N (number of samples), then channels, then H, W.
# We can flatten and find unique values, or just take the value from a known pixel.
x_channel0_values = X_combined[:, 0, 0, 0] # Take one pixel value from each sample's channel 0
x_channel0_max = np.max(x_channel0_values)
X_normalized[:, 0, :, :] = X_combined[:, 0, :, :] / x_channel0_max
print(f"X Channel 0 (Reynolds): Normalized by dividing by max value: {x_channel0_max}")

# Channel 1 (Geometry Mask) - values between approx -0.29 and 1.04
# Normalize to [0, 1] using min-max scaling
x_channel1_min = X_combined[:, 1, :, :].min()
x_channel1_max = X_combined[:, 1, :, :].max()
X_normalized[:, 1, :, :] = (X_combined[:, 1, :, :] - x_channel1_min) / (x_channel1_max - x_channel1_min)
print(f"X Channel 1 (Geometry Mask): Min-max normalized to [0,1] (orig_min={x_channel1_min:.4f}, orig_max={x_channel1_max:.4f})")

# Channel 2 (SDF - but observed as a binary mask [0, 255])
# Normalize to [0, 1] by dividing by 255
x_channel2_min = X_combined[:, 2, :, :].min()
x_channel2_max = X_combined[:, 2, :, :].max()
X_normalized[:, 2, :, :] = X_combined[:, 2, :, :] / x_channel2_max # Assuming max is 255
print(f"X Channel 2 (SDF/Mask): Normalized to [0,1] (orig_min={x_channel2_min:.4f}, orig_max={x_channel2_max:.4f})")


# --- 2. Normalize Y_combined (u, v, p, T) ---
print("\nNormalizing Y_combined...")
Y_normalized = np.copy(Y_combined).astype(np.float32)

y_normalizer_params = {}

for i in range(Y_combined.shape[1]): # Iterate through each channel
    channel_data = Y_combined[:, i, :, :]
    mean_val = channel_data.mean()
    std_val = channel_data.std()

    # Store parameters
    y_normalizer_params[f'channel_{i}'] = {'mean': mean_val, 'std': std_val}

    # Apply standardization
    Y_normalized[:, i, :, :] = (channel_data - mean_val) / std_val
    print(f"Y Channel {i}: Standardized (mean={mean_val:.4f}, std={std_val:.4f})")

# --- 3. Convert to PyTorch Tensors ---
X_tensor = torch.from_numpy(X_normalized).float()
Y_tensor = torch.from_numpy(Y_normalized).float()

print(f"\nConverted X_normalized to PyTorch tensor with shape: {X_tensor.shape} and dtype: {X_tensor.dtype}")
print(f"Converted Y_normalized to PyTorch tensor with shape: {Y_tensor.shape} and dtype: {Y_tensor.dtype}")

print("\nNormalization parameters for X:")
print(f"  X Channel 0 max: {x_channel0_max}")
print(f"  X Channel 1 min: {x_channel1_min}, max: {x_channel1_max}")
print(f"  X Channel 2 min: {x_channel2_min}, max: {x_channel2_max}")

print("\nNormalization parameters for Y:")
for ch, params in y_normalizer_params.items():
    print(f"  {ch}: Mean={params['mean']:.4f}, Std={params['std']:.4f}")
````

## Define FNO Model Architecture

### Subtask:
Implement the Fourier Neural Operator (FNO) model architecture using PyTorch, including the necessary layers such as Fourier layers and linear transformations.


````python
import torch
import torch.nn as nn
import torch.fft as fft

class SpectralConv2d(nn.Module):
    def __init__(self, in_channels, out_channels, modes1, modes2):
        super(SpectralConv2d, self).__init__()
        self.in_channels = in_channels
        self.out_channels = out_channels
        self.modes1 = modes1 # Number of Fourier modes to retain for height
        self.modes2 = modes2 # Number of Fourier modes to retain for width (for rfft2, this implies W//2 + 1 modes)

        self.scale = (1 / (in_channels * out_channels))
        # Learnable complex weights for the low-frequency modes
        # Weights for the top-left frequency quadrant
        self.weights1 = nn.Parameter(self.scale * torch.rand(in_channels, out_channels, self.modes1, self.modes2, dtype=torch.cfloat))
        # Weights for the bottom-left frequency quadrant (corresponding to negative frequencies in the first spatial dimension)
        self.weights2 = nn.Parameter(self.scale * torch.rand(in_channels, out_channels, self.modes1, self.modes2, dtype=torch.cfloat))

        # Linear layer for skip connection (spatial domain transformation)
        # A 1x1 convolution (Conv2d) is typically used in 2D FNOs for this purpose,
        # acting as a channel-wise linear projection in the spatial domain.
        self.linear_skip = nn.Conv2d(in_channels, out_channels, kernel_size=1)

    def forward(self, x):
        batchsize = x.shape[0]
        height = x.shape[-2]
        width = x.shape[-1]

        # 1. Compute 2D Real Fast Fourier Transform (RFFT)
        # x_ft has shape (batch, in_channels, H, W//2 + 1) for real input
        x_ft = torch.fft.rfft2(x)

        # Create an output tensor to store the filtered Fourier modes
        # It will have the same dimensions as x_ft, but with out_channels
        out_ft = torch.zeros(batchsize, self.out_channels, height, width // 2 + 1, dtype=torch.cfloat, device=x.device)

        # 2. Multiply the lower-frequency modes with the learnable complex weights
        # Apply weights to the top-left and bottom-left frequency quadrants of the RFFT output
        # These slices represent the low-frequency modes to be modified.
        out_ft[:, :, :self.modes1, :self.modes2] = \
            torch.einsum("bixy,ioxy->boxy", x_ft[:, :, :self.modes1, :self.modes2], self.weights1)

        out_ft[:, :, -self.modes1:, :self.modes2] = \
            torch.einsum("bixy,ioxy->boxy", x_ft[:, :, -self.modes1:, :self.modes2], self.weights2)

        # 3. Perform the inverse 2D Fast Fourier Transform (IFFT)
        # The 's' argument ensures the output spatial dimensions match the original input
        x_spectral_out = torch.fft.irfft2(out_ft, s=(height, width))

        # 4. Add the result from a linear skip connection (spatial domain transformation)
        x_skip_out = self.linear_skip(x)

        return x_spectral_out + x_skip_out

print("SpectralConv2d class defined.")
````

               ↓ Apply low-freq learnable weights
           Low-freq filtered spectrum
 --iFFT--> Spatial representation (global update)
      +    1×1 conv (local update)
------------------------------------------
               Output


Now that the `SpectralConv2d` class is defined, I will proceed to implement the main `FNO2d` class, which will integrate these spectral convolution blocks with lifting and projection layers to form the complete FNO architecture.

**Reasoning**:
Now that the `SpectralConv2d` class is defined, I will implement the main `FNO2d` class. This class will combine the `SpectralConv2d` layers with lifting and projection linear layers, and activation functions to form the complete Fourier Neural Operator architecture, handling the input and output channel dimensions.



````python
class FNO2d(nn.Module):
    def __init__(self, in_channels, out_channels, width, modes1, modes2):
        super(FNO2d, self).__init__()

        self.in_channels = in_channels
        self.out_channels = out_channels
        self.width = width
        self.modes1 = modes1
        self.modes2 = modes2

        # Lifting layer: projects input channels to 'width' dimension
        self.lifting = nn.Linear(in_channels, self.width) # Input is (..., C)

        # Spectral Convolution layers
        # Each SpectralConv2d block increases the feature dimension from 'width' to 'width'
        self.spectral_conv1 = SpectralConv2d(self.width, self.width, modes1, modes2)
        self.spectral_conv2 = SpectralConv2d(self.width, self.width, modes1, modes2)
        self.spectral_conv3 = SpectralConv2d(self.width, self.width, modes1, modes2)
        self.spectral_conv4 = SpectralConv2d(self.width, self.width, modes1, modes2)

        # Linear layers for transformations between spectral blocks
        # These are 1x1 convolutions after channel-first (B, C, H, W) permute
        self.linear1 = nn.Conv1d(self.width, self.width, 1)
        self.linear2 = nn.Conv1d(self.width, self.width, 1)
        self.linear3 = nn.Conv1d(self.width, self.width, 1)

        # Projection layer: maps 'width' dimension back to output channels
        self.projection = nn.Linear(self.width, out_channels) # Output is (..., C)

        # Activation function
        self.activation = nn.GELU()

    def forward(self, x):
        # x shape: (batch_size, in_channels, height, width)

        # 1. Lifting layer
        # Permute to (batch_size, height, width, in_channels) for Linear layer
        x = x.permute(0, 2, 3, 1)
        x = self.lifting(x) # (batch_size, height, width, self.width)
        # Permute back to (batch_size, self.width, height, width) for Conv layers
        x = x.permute(0, 3, 1, 2)

        # 2. Sequence of SpectralConv2d blocks with activation and linear transformations
        x1 = self.spectral_conv1(x)
        x2 = self.linear1(x.view(x.shape[0], x.shape[1], -1)).view(x.shape) # Apply 1x1 conv on flattened H*W, then reshape
        x = self.activation(x1 + x2)

        x1 = self.spectral_conv2(x)
        x2 = self.linear2(x.view(x.shape[0], x.shape[1], -1)).view(x.shape)
        x = self.activation(x1 + x2)

        x1 = self.spectral_conv3(x)
        x2 = self.linear3(x.view(x.shape[0], x.shape[1], -1)).view(x.shape)
        x = self.activation(x1 + x2)

        # 3. Projection layer
        # Permute to (batch_size, height, width, self.width) for Linear layer
        x = x.permute(0, 2, 3, 1)
        x = self.projection(x) # (batch_size, height, width, out_channels)
        # Permute back to (batch_size, out_channels, height, width) for final output consistency
        x = x.permute(0, 3, 1, 2)

        return x

print("FNO2d class defined.")
````

### `SpectralConv2d` Class Explanation

The `SpectralConv2d` class is the core building block of the FNO model, responsible for performing convolution in the Fourier (frequency) domain. Here's a breakdown:

1.  **Initialization (`__init__`)**:
    *   `in_channels`, `out_channels`: Number of input and output feature channels.
    *   `modes1`, `modes2`: These define the number of low-frequency Fourier modes to retain along the height and width dimensions, respectively. High-frequency modes are truncated, which is a key characteristic of FNO for efficiency and generalization.
    *   `weights1`, `weights2`: These are `nn.Parameter` objects containing learnable complex weights. These weights are applied to the retained low-frequency modes in the Fourier domain. `weights1` typically handles the top-left quadrant of the frequency spectrum, and `weights2` handles the bottom-left quadrant (for real-valued input using `rfft2`).
    *   `linear_skip`: This is a standard `nn.Conv2d` layer with a 1x1 kernel. It acts as a skip connection in the spatial domain, allowing information to bypass the Fourier transform operations. This helps the model retain local information and can improve stability.

2.  **Forward Pass (`forward`)**:
    *   **Real Fast Fourier Transform (`torch.fft.rfft2(x)`)**: The input tensor `x` (which is in the spatial domain) is transformed into the frequency domain. `rfft2` is used because the input `x` is real, which makes the Fourier transform symmetric and reduces computation by only computing half of the spectrum.
    *   **Mode Multiplication**: The transformed Fourier coefficients (`x_ft`) are then selectively multiplied by the `weights1` and `weights2` parameters. Only the low-frequency modes (defined by `modes1` and `modes2`) are retained and transformed; higher frequencies are implicitly zeroed out. The `torch.einsum` operation is used for efficient batched multiplication of the input Fourier coefficients with the learnable weights.
    *   **Inverse Real Fast Fourier Transform (`torch.fft.irfft2(out_ft, s=(height, width))`)**: The modified Fourier coefficients (`out_ft`) are transformed back into the spatial domain using the inverse FFT. The `s` argument ensures the output spatial dimensions match the original input.
    *   **Skip Connection**: The result from the inverse FFT (`x_spectral_out`) is added to the output of the spatial `linear_skip` connection (`x_skip_out`). This combines the global information captured by the Fourier convolution with local information from the linear skip connection.

In essence, `SpectralConv2d` learns to apply frequency-dependent filters to the input data, effectively performing convolutions in the Fourier domain, and combines this with a direct spatial transformation.

### `FNO2d` Class Explanation

The `FNO2d` class combines multiple `SpectralConv2d` layers with linear transformations to create the full Fourier Neural Operator architecture. Here's how it's structured:

1.  **Initialization (`__init__`)**:
    *   `in_channels`, `out_channels`: Number of input and output channels for the entire FNO model.
    *   `width`: This is a crucial hyperparameter representing the hidden dimension or width of the FNO. The input features are 'lifted' to this dimension, and all intermediate spectral convolutions operate within this dimension.
    *   `modes1`, `modes2`: Passed directly to the `SpectralConv2d` layers.
    *   **`lifting = nn.Linear(in_channels, self.width)`**: This is the initial

`lifting` layer. It maps the original input channels to the higher-dimensional `width` space where the spectral operations occur. Since `nn.Linear` expects a channel-last format (e.g., `B, H, W, C`), the input is permuted before and after this operation.
    *   **`spectral_conv1` to `spectral_conv4`**: Four instances of the `SpectralConv2d` layer are defined. Each takes `self.width` as both `in_channels` and `out_channels`, maintaining the hidden dimension.
    *   **`linear1`, `linear2`, `linear3`**: These are `nn.Conv1d` layers with a kernel size of 1. In the context of a 2D FNO, when applied after permuting the input, they effectively act as 1x1 convolutions on the feature maps (`width` dimension). These serve as residual connections within the spectral blocks, helping with information flow and preventing vanishing gradients.
    *   **`projection = nn.Linear(self.width, out_channels)`**: This is the final layer that projects the features from the hidden `width` dimension back to the desired `out_channels` of the model's output.
    *   **`activation = nn.GELU()`**: The GELU (Gaussian Error Linear Unit) activation function is used after each spectral block to introduce non-linearity.

2.  **Forward Pass (`forward`)**:
    *   **Lifting**: The input `x` is first permuted to `(batch_size, height, width, in_channels)` to be compatible with the `nn.Linear` layer. It's then passed through the `lifting` layer and permuted back to `(batch_size, self.width, height, width)` for the convolutional layers.
    *   **Spectral Blocks**: The model then proceeds through a sequence of four spectral convolution blocks. Each block consists of:
        1.  A `SpectralConv2d` layer (`x1`).
        2.  A linear transformation (`x2`) applied via `nn.Conv1d` (effectively a 1x1 convolution across the spatial dimensions after flattening and reshaping).
        3.  The output is the `activation` of `x1 + x2`, forming a residual-like connection.
    *   **Projection**: Finally, the output of the last spectral block is again permuted to `(batch_size, height, width, self.width)` and passed through the `projection` layer to obtain the final output of shape `(batch_size, out_channels, height, width)`.
